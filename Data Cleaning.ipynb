```python
import pandas as pd

df = pd.read_csv("Data Cleaning 12.08.csv")

print(df.head())
```

```python
print(df.shape)
```

```python
print(df.columns.tolist())
```

```python
print("Total planets:", len(df))
print("Average mass:", df["mass"].mean())
print("Average radius:", df["radius"].mean())
print("Highest mass:", df["mass"].max())
print("Lowest mass:", df["mass"].min())
```

```python
highest = df.loc[df["mass"].idxmax(), "name"]
lowest = df.loc[df["mass"].idxmin(), "name"]

print("Planet with highest mass:", highest)
print("Planet with lowest mass:", lowest)
```

```python
top_10 = df.nlargest(10, "mass")[["name", "mass"]]

print(top_10)
```

```python
highest_planet = df.loc[df["mass"].idxmax(), "name"]

lowest_planet = df.loc[df["mass"].idxmin(), "name"]

print("Highest mass planet:", highest_planet)
print("Highest mass:", df["mass"].max())

print("Lowest mass planet:", lowest_planet)
print("Lowest mass:", df["mass"].min())
```

```python
top_10 = df.nlargest(10, "mass")[["name", "mass"]]

print(top_10)
```

```python

total_planets = len(df)
average_mass = df["mass"].mean()
average_radius = df["radius"].mean()
highest_mass = df["mass"].max()
lowest_mass = df["mass"].min()

print("===== DASHBOARD STATISTICS =====")
print("Total Planets:", total_planets)
print("Average Mass:", average_mass)
print("Average Radius:", average_radius)
print("Highest Mass:", highest_mass)
print("Lowest Mass:", lowest_mass)
print("Highest Mass Planet:", highest_planet)
print("Lowest Mass Planet:", lowest_planet)
```

```python
import matplotlib.pyplot as plt

status_count = df["planet_status"].value_counts()

print(status_count)
```

```python
plt.figure(figsize=(8, 5))

status_count.plot(kind="bar")

plt.title("Number of Planets by Planet Status")
plt.xlabel("Planet Status")
plt.ylabel("Number of Planets")

plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
top_10 = df.nlargest(10, "mass")[["name", "mass"]]

plt.figure(figsize=(10, 6))

plt.bar(top_10["name"], top_10["mass"])

plt.title("Top 10 Planets by Mass")
plt.xlabel("Planet")
plt.ylabel("Mass")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(df["mass"], df["radius"])
plt.title("Relationship Between Planet Mass and Radius")
plt.xlabel("Planet Mass")
plt.ylabel("Planet Radius")
plt.tight_layout()
plt.show()
```

```python
top_10.to_csv("Top_10_Planets_by_Mass.csv", index=False)
```

```python
summary = pd.DataFrame({
    "Metric": [
        "Total Planets",
        "Average Mass",
        "Average Radius",
        "Highest Mass",
        "Lowest Mass",
        "Highest Mass Planet",
        "Lowest Mass Planet"
    ],
    "Value": [
        total_planets,
        average_mass,
        average_radius,
        highest_mass,
        lowest_mass,
        highest_planet,
        lowest_planet
    ]
})

print(summary)
```

```python
summary.to_csv("Planet_Dashboard_Summary.csv", index=False)
```
